# Question 5: Calculate the sum of digits in number 12345

```python
num = 12345
temp = num
sum = 0
while temp>0:
    last = temp%10
    sum = sum+last
    temp = temp//10
print(sum)    



```

# Question 6: Check if 97 is a prime number

```python
num = 97
if num<=1:
    print("Not prime number")
else:
    for i in range(2,num):
        if num%i == 0:
            print("Not prime number")
            break
    print("prime number")    


```

# Question 7: Find the factorial of 8

```python
num = 8
fact = 1
for i in range(1,num+1):
    fact *= i
print(fact)    
```

# Question 8: Calculate the average of numbers: 15, 23, 31, 42, 56

```python
num = [15,23,31,42,56]
avg = sum(num)/len(num)
print(avg)
```

# Question 9: Find the greatest common divisor (GCD) of 48 and 36

```python
num1 = 48
num2 = 36
while num2:
    num1, num2 = num2, num1%num2
print("GCD=", num1)    
```
