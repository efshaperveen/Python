
# Question 11: Find the most frequent character in "mississippi"

```python
text = "mississippi"
count = {}
for ch in text:
    count[ch] = count.get(ch,0) + 1
most_frequent = max(count, key = count.get)
print(most_frequent)
```

```python
s = "mississippi"
max_char = ""
max_count = 0
for ch in s:
    if s.count(ch)> max_count:
        max_count = s.count(ch)
        max_char = ch
print(f"Most frequent character : {max_char}")
```

# Question 41: Check if string contains balanced parentheses: "((()))"

```python
s = "((()))"

count = 0

for ch in s:
    if ch == '(':
        count += 1
    elif ch == ')':
        count -= 1

    if count < 0:
        break

if count == 0:
    print("Balanced")
else:
    print("Not Balanced")
```

# Question 42: Convert "hello world" to Morse code

```python
s = "hello world"

morse = {
    'h': '....', 'e': '.', 'l': '.-..', 'o': '---',
    'w': '.--', 'r': '.-.', 'd': '-..'
}

words = s.split()
result = []

for word in words:
    code = ""
    for ch in word:
        code += morse[ch] + " "
    result.append(code.strip())

print(" / ".join(result))
```

# Question 43: Find the longest common substring between "programming" and "grammar"

```python
s1 = "programming"
s2 = "grammar"

longest = ""

for i in range(len(s1)):
    for j in range(i + 1, len(s1) + 1):
        sub = s1[i:j]

        if sub in s2 and len(sub) > len(longest):
            longest = sub

print(longest)
```

# Question 44: Check if string is a valid URL: "https://www.google.com"

# Question 46: Convert "hello world" to Pig Latin


```python
s = "hello world"

words = s.split()
result = []

for word in words:
    if word[0] in "aeiou":
        result.append(word + "way")
    else:
        result.append(word[1:] + word[0] + "ay")

print(" ".join(result))
```


# Question 50: Check if string is a valid credit card number: "4532015112830366"


```python
s = "4532015112830366"

if not s.isdigit():
    print("Invalid card number")
else:
    digits = [int(ch) for ch in s]
    checksum = 0

    for i in range(len(digits) - 1, -1, -1):
        digit = digits[i]

        if (len(digits) - 1 - i) % 2 == 1:
            digit *= 2

            if digit > 9:
                digit -= 9

        checksum += digit

    if checksum % 10 == 0:
        print("Valid card number")
    else:
        print("Invalid card number")
```
